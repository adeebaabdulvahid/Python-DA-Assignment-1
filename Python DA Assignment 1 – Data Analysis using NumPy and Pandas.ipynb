### Pandas Series

1. Creating Pandas Series

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

2. Indexing and Slicing

a) Access the mark of the 1st rank student using integer position

```python
print("1st Rank Mark:", marks.iloc[0])
```

b) Retrieve the top 3 ranks using loc

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

c) Retrieve the mark of the 3rd rank student using iloc

```python
print("3rd Rank Mark:", marks.iloc[2])
```

d) Filter students who scored more than 90 marks

```python
print(marks[marks > 90])
```

3. Manipulating Series

a) Change the 1st rank student's mark to 100

```python
marks.loc['Rank1'] = 100

print(marks)
```

b) Remove the last rank from the Series

```python
marks = marks.drop('Rank5')

print(marks)
```

c) Calculate CGPA by dividing each mark by 10

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

Part 3: Pandas DataFrame

1. Creating Pandas DataFrame

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

2. Data Exploration

a) Display DataFrame and its basic information

```python
print("Transactions DataFrame:")
print(transactions)

print("\nFirst 5 Rows:")
print(transactions.head())

print("\nLast 5 Rows:")
print(transactions.tail())

print("\nShape:")
print(transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)
```

b) Display only ProductCategory and Amount columns

```python
print(transactions[['ProductCategory', 'Amount']])
```

c) Retrieve the last 3 columns using iloc

```python
print(transactions.iloc[:, -3:])
```

d)Filter rows where Region is North and Amount is greater than 200

```python
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(result)
```

e) Find the value counts for ProductCategory

```python
print(transactions['ProductCategory'].value_counts())
```

f) Find the unique values in Region

```python
print(transactions['Region'].unique())
```

g) Group by Region and find the mean amount

```python
print(transactions.groupby('Region')['Amount'].mean())
```

3. Manipulating the DataFrame

a) Modify the Amount of TransactionID 102 to 165

```python
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print(transactions)
```

b) Add a new Discount column by calculating 10% of Amount

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

c) Remove the row with TransactionID 109

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

d) Delete the Discount column

```python
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```
